## Задание 1

```python
import math
import numpy as np
import matplotlib.pyplot as plt

# Данные
a = 1.75
b = 1.21
c = 0.041

# Функция из уравнения
def f(x):
    # Подставляем формулу варианта, где x выступает в роли переменной a
    return (x - math.sin(b)) / (b**2 + 6 * c)

print("Графический метод")

# Строим графики (возьмем зависимость Z от переменной a)
x = np.linspace(0, 4, 200) # можно взять от 0 до 4
y1 = (x - np.sin(b)) / (b**2 + 6 * c) # расчет функции
y2 = np.zeros_like(x) # линия нуля для поиска пересечения

plt.plot(x, y1, label='(a - sin(b)) / (b² + 6c)')
plt.plot(x, y2, label='0', color='red', linestyle='--')
plt.grid(True)
plt.legend()
plt.title('7 вариант: Расчет и поиск точки пересечения')
plt.xlabel('x (в роли a)')
plt.ylabel('Z')
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.show()

# Находим интервалы (разность меняет знак)
print("\nИнтервалы, содержащие корни:")
for i in range(len(x)-1):
    if (y1[i]) * (y1[i+1]) < 0:
        print(f"Корень на [{x[i]:.2f}; {x[i+1]:.2f}]")

# Расчет конкретного значения Z для заданного a=1.75
z_final = f(a)
print(f"\nИтоговое значение Z для a={a}: {z_final:.5f}")
```

## Задание 2

```python
import math

# Данные для 7 варианта
b_val = 1.21
c_val = 0.041

# Функция из 7 варианта, приведенная к виду f(x) = 0
# x выступает в роли переменной a
def f(x):
    # Формула: (x - sin(b)) / (b^2 + 6c)
    return (x - math.sin(b_val)) / (b_val**2 + 6 * c_val)

print("Уравнение: (x - sin(b)) / (b² + 6c) = 0")

# Интервал подбираем так, чтобы f(a) и f(b) имели разные знаки
# Корень x = sin(1.21) ≈ 0.935, поэтому возьмем интервал [0.9; 1.0]
a, b, eps = 0.9, 1.0, 0.001

print(f"Начальный интервал: [{a}; {b}]")
print(f"Точность: {eps}")

# Для подсчета итераций
iteration = 1

while b - a > eps:
    c = (a + b) / 2
    fc = f(c)
    print(f"Итерация {iteration}: a={a:.6f}, b={b:.6f}, c={c:.6f}, f(c)={fc:.6f}")
    
    # Стандартная логика метода половинного деления
    if f(a) * fc <= 0:
        b = c
    else:
        a = c
    iteration += 1

root = (a + b) / 2
print(f"\nКорень: {root:.6f}")
print(f"Проверка: f({root:.6f}) = {f(root):.6f}")
print(f"Число итераций: {iteration-1}")
print(f"Достигнутая точность: {b-a:.6f}")

# Проверка подстановкой (сравнение с теоретическим корнем sin(b))
theoretical = math.sin(b_val)
print(f"\nПроверка через теорию:")
print(f"Найденный x = {root:.6f}")
print(f"Ожидаемый x (sin({b_val})) = {theoretical:.6f}")
print(f"Разность: {abs(root - theoretical):.6f}")
```

```python
import math

# Начальное значение (выбираем 1.1, так как корень рядом)
x = 1.1  
eps = 0.001

while True:
    # Итерационная формула: x = 1 / sin(x)
    x_new = 1 / math.sin(x)
    
    if abs(x_new - x) < eps:
        break
    x = x_new

print(f'Корень: {x:.4f}')
# Проверка для уравнения x * sin(x) = 1
print(f'Проверка: {x:.4f} * sin({x:.4f}) = {x * math.sin(x):.4f}')
print(f'Целевое значение: 1.0000')
```

```python
import numpy as np
import matplotlib.pyplot as plt
import math


# Строим графики двух функций
x = np.linspace(0, 10, 400)   # возьмем интервал от 0 до 10, чтобы увидеть несколько корней
y1 = x * np.sin(x)            # левая часть уравнения: x * sin(x)
y2 = np.ones_like(x)          # правая часть уравнения: 1

plt.figure(figsize=(10, 6))
plt.plot(x, y1, label='y = x * sin(x)')
plt.plot(x, y2, label='y = 1', color='red', linestyle='--')

plt.grid(True)
plt.legend()
plt.title('Отделение корней уравнения x * sin(x) = 1')
plt.xlabel('x')
plt.ylabel('y')

# Рисуем оси
plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)
plt.show()

# Находим интервалы, где функции пересекаются
print("\nИнтервалы, содержащие корни:")
for i in range(len(x)-1):
    # Если разность (y1 - y2) меняет знак, значит линии пересеклись
    if (y1[i] - y2[i]) * (y1[i+1] - y2[i+1]) < 0:
        print(f"Корень на [{x[i]:.2f}; {x[i+1]:.2f}]")
```

## Задание 2

```python
import math

def f(x):
    return x * math.sin(x) - 1

# Интервал, где f(1.0) < 0, а f(1.2) > 0
a, b = 1.0, 1.2 
eps = 0.001

print('n     a         b         x         f(x)      x_n - x_{n-1}')

x_prev = a
n = 1
while True:
    # Формула метода хорд
    x = a - f(a) * (b - a) / (f(b) - f(a))
    diff = abs(x - x_prev)
    
    print(f'{n:2d}   {a:.6f} {b:.6f} {x:.6f} {f(x):.6f} {diff:.6f}')
    
    # Условие остановки: достигнута точность по функции или по X
    if abs(f(x)) < eps or diff < eps:
        print(f'\nКорень: x = {x:.6f}')
        break
        
    # Уточнение интервала
    if f(a) * f(x) < 0:
        b = x
    else:
        a = x
    
    x_prev = x
    n += 1

# Проверка подстановкой в исходное уравнение x * sin(x) = 1
left_side = x * math.sin(x)
print(f'\nПроверка исходного уравнения:')
print(f'Левая часть (x * sin(x)) = {left_side:.6f}')
print(f'Правая часть (1) = 1.000000')
print(f'Разность: {left_side - 1:.6f}')
```

## Задание 3

```python
import math

def f(x):
    return x * math.sin(x) - 1

# Производная функции: sin(x) + x*cos(x)
def f_derivative(x):
    return math.sin(x) + x * math.cos(x)

# Начальное приближение (возьмем 1.1 из графического метода)
x = 1.1  
eps = 0.001

print('n     x_n        f(x_n)      x_n - x_{n-1}')

n = 1
while True:
    # Формула Ньютона: x_new = x - f(x) / f'(x)
    x_new = x - f(x) / f_derivative(x)
    diff = abs(x_new - x)
    
    print(f'{n:2d}   {x:.6f}   {f(x):.6f}   {diff:.6f}')
    
    if abs(f(x_new)) < eps or diff < eps:
        x = x_new # сохраняем последний результат
        print(f'\nКорень: x = {x:.6f}')
        break
    
    x = x_new
    n += 1

left_side = x * math.sin(x)
print(f'Проверка: {x:.6f} * sin({x:.6f}) = {left_side:.6f}')
print(f'Разность с единицей: {abs(left_side - 1):.6f}')
```
